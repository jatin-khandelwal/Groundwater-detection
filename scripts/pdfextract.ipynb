```python
import pdfplumber
import pandas as pd

data = []
header = None

with pdfplumber.open(r"C:\Users\Jatin Khandelwal\Downloads\RAJASTHAN.pdf") as hi:

    for page in hi.pages:
        for table in page.extract_tables():

            if header is None:
                header = table[0]

            for row in table[1:]:
                if row[5] == "Rajasthan":
                    data.append(row)

df = pd.DataFrame(data, columns=header)

df.to_excel("RAJASTHAN_aug_2021_nov_2021_jan_2022.xlsx", index=False)
```

```python
import pandas as pd

# Step 1: Load all 4 files into separate DataFrames
df1 = pd.read_excel("RAJASTHAN_aug_2021_nov_2021_jan_2022.xlsx")
df2 = pd.read_excel("RAJASTHAN_aug_2022.xlsx")
df3 = pd.read_excel("RAJASTHAN_may_2022.xlsx")
df4 = pd.read_excel("RAJASTHAN_nov_2022.xlsx")

# Step 2: Fix the column name mismatch
# may_2022 and nov_2022 have "WLS_WTR_LEVE" (one letter short) instead of "WLS_WTR_LEVEL"
df3 = df3.rename(columns={"WLS_WTR_LEVE": "WLS_WTR_LEVEL"})
df4 = df4.rename(columns={"WLS_WTR_LEVE": "WLS_WTR_LEVEL"})

# Step 3: Combine all 4 into one dataset
combined = pd.concat([df1, df2, df3, df4], ignore_index=True)

# Step 4: Quick sanity checks
print("Final shape (rows, columns):", combined.shape)
print("Columns:", list(combined.columns))
print("Any missing water level values?", combined["WLS_WTR_LEVEL"].isna().sum())

# Step 5: Save the merged dataset
combined.to_excel("rajasthan_groundwater_merged.xlsx", index=False)
```
